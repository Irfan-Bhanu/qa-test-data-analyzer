```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.service import Service
from webdriver_manager.chrome import ChromeDriverManager
from selenium.webdriver.common.keys import Keys
import time
import csv
import logging

logging.basicConfig(
    filename="test_log.log",
    level=logging.INFO,
    format = "%(asctime)s - %(message)s"
)

def setup_driver():
    driver = webdriver.Chrome(service=Service(ChromeDriverManager().install()))
    driver.maximize_window()
    return driver

def login_function(driver,username,password):
    driver.get("https://www.saucedemo.com")
    driver.find_element(By.ID,"user-name").clear()
    driver.find_element(By.ID,"user-name").send_keys(username)# this is same as JAVA , but ID should be in caps
    #password
    driver.find_element(By.ID,"password").clear()
    driver.find_element(By.ID,"password").send_keys(password)
    #login button 
    driver.find_element(By.NAME,"login-button").click()
    time.sleep(3)

def test_case_website_validation(driver,username):
    if "inventory" in driver.current_url:
            logging.info(f"{username} → Login Passed")
            driver.find_element(By.ID,"add-to-cart-sauce-labs-backpack").click()
            driver.find_element(By.CLASS_NAME,"shopping_cart_link").click()# note the spelling is class_name
            item= driver.find_element(By.CLASS_NAME,"inventory_item_name").text #note text is not a function
            logging.info(f"Product found in cart:,{item}")
    else:
        logging.info(f"{username} → Login failed")

    logging.info("----------------------")

def main():
    driver = setup_driver()

    with open(r"C:\Users\bhanu\Downloads\login_data.csv", mode='r') as file:
        reader = csv.DictReader(file)

        for row in reader:
            username = row["username"]
            password = row["password"]

            login_function(driver, username, password)
            test_case_website_validation(driver, username)

    driver.quit()


# Run script
main()
    
```

```python
#Read data from excel

from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.service import Service
from webdriver_manager.chrome import ChromeDriverManager
from selenium.webdriver.common.keys import Keys
import time
import csv
import logging
import pandas as pd


logging.basicConfig(
    filename="test_log.log",
    level=logging.INFO,
    format = "%(asctime)s - %(message)s"
)

def setup_driver():
    driver = webdriver.Chrome(service=Service(ChromeDriverManager().install()))
    driver.maximize_window()
    return driver

def login_function(driver,username,password):
    driver.get("https://www.saucedemo.com")
    driver.find_element(By.ID,"user-name").clear()
    driver.find_element(By.ID,"user-name").send_keys(username)# this is same as JAVA , but ID should be in caps
    #password
    driver.find_element(By.ID,"password").clear()
    driver.find_element(By.ID,"password").send_keys(password)
    #login button 
    driver.find_element(By.NAME,"login-button").click()
    time.sleep(3)

def test_case_website_validation(driver,username,test_results):
    
    if "inventory" in driver.current_url:
            status="PASS"
            logging.info(f"{username} → Login Passed")
            driver.find_element(By.ID,"add-to-cart-sauce-labs-backpack").click()
            driver.find_element(By.CLASS_NAME,"shopping_cart_link").click()# note the spelling is class_name
            item= driver.find_element(By.CLASS_NAME,"inventory_item_name").text #note text is not a function
            logging.info(f"Product found in cart:,{item}")
    else:
        status="FAIL"
        logging.info(f"{username} → Login failed")

    logging.info("----------------------")
    test_results.append({
        "username": username,
        "status":status})
    


def main():
    test_results=[]
    driver = setup_driver()

    data=pd.read_excel(r"C:\Users\bhanu\OneDrive\Desktop\login_data_excel.xlsx")
    

    for index,row in data.iterrows():
            username = row["username"]
            password = row["password"]

            login_function(driver, username, password)
            test_case_website_validation(driver, username,test_results)
        
    print("\n------ FINAL TEST REPORT ------")
    
    pass_count = sum(1 for r in test_results if r["status"] == "PASS")
    fail_count = sum(1 for r in test_results if r["status"] == "FAIL")
    
    print(f"Total Passed: {pass_count}")
    print(f"Total Failed: {fail_count}")
    
    for r in test_results:
        print(f"{r['username']} → {r['status']}")

    driver.quit()


# Run script
main()
    
```
