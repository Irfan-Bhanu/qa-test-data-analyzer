```python
#data driven testing

from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.service import Service
from webdriver_manager.chrome import ChromeDriverManager
from selenium.webdriver.common.keys import Keys
import time
import csv

driver = webdriver.Chrome(service=Service(ChromeDriverManager().install()))


#read username and pwd from excel

with open(r"C:\Users\bhanu\Downloads\login_data.csv",mode='r') as file:
    reader = csv.DictReader(file)

    for i in reader:
        username=i['username']
        password=i['password']
        
        url= driver.get("https://www.saucedemo.com")
        driver.maximize_window()

#find element- 
    #username
        driver.find_element(By.ID,"user-name").clear()
        driver.find_element(By.ID,"user-name").send_keys(username)# this is same as JAVA , but ID should be in caps
        #password
        driver.find_element(By.ID,"password").clear()
        driver.find_element(By.ID,"password").send_keys(password)
        #login button 
        driver.find_element(By.NAME,"login-button").click()
        time.sleep(3)
        #get title
        #print("New Page title is ",driver.title) #title is not a function
        
        
        '''searchbox=driver.find_element(By.NAME,"q")
        searchbox.send_keys("Selenium Python")
        searchbox.send_keys(Keys.ENTER)
        driver.find_element(By.CSS_SELECTOR, "h3")
        print("New Page Title:" ,driver.title)'''
        
        
        time.sleep(2)
        
        #print(f"logged in with {username}")
        
        #test case- Login → Add product → Open cart → Verify product
        if "inventory" in driver.current_url:
            print(f"{username} → Login Passed")
            driver.find_element(By.ID,"add-to-cart-sauce-labs-backpack").click()
            driver.find_element(By.CLASS_NAME,"shopping_cart_link").click()# note the spelling is class_name
            item= driver.find_element(By.CLASS_NAME,"inventory_item_name").text #note text is not a function
            print("Product found in cart:",item)
        else:
            print(f"{username} → Login failed")

        print("----------------------")

driver.quit()
```
