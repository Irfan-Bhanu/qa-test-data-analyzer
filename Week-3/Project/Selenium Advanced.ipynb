```python
#1.dropdown selection
from selenium import webdriver
from selenium.webdriver.support.ui import Select
from selenium.webdriver.common.by import By

#initialise driver
driver=webdriver.Chrome()
driver.maximize_window()
#navigate to url
url=driver.get("https://www.w3schools.com/tags/tryit.asp?filename=tryhtml_select")
print("Title of the page - ",driver.title)

#select the dropdown

driver.switch_to.frame(driver.find_element(By.ID,"iframeResult"))#switch_to is same ass java
dropdown=driver.find_element(By.ID,"cars")


select= Select(dropdown)

#select by visible text
select.select_by_visible_text("Opel")
#print the selected value
selected= select.first_selected_option.text
print("selected text is",selected)

#selectbyvalue
select.select_by_value("saab")
selected= select.first_selected_option.text
print("selected text is",selected)

#selectbyindex
select.select_by_index(0)
selected= select.first_selected_option.text
print("selected text is",selected)

#close the browser
driver.quit()
```

```python
#3. Handling Alerts

from selenium import webdriver
from selenium.webdriver.common.by import By
import time
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC

#initialise driver
driver=webdriver.Chrome()
driver.maximize_window()
#navigate to url
driver.get("https://www.w3schools.com/js/tryit.asp?filename=tryjs_prompt")
print("Title of the page - ",driver.title)

driver.switch_to.frame("iframeResult")
driver.find_element(By.XPATH,"//button[text()='Try it']").click()

wait = WebDriverWait(driver, 10)
alert = wait.until(EC.alert_is_present())#explicit wait

alert=driver.switch_to.alert
print("Alert text:", alert.text)
alert.dismiss()

text_displayed=driver.find_element(By.ID,"demo").text
print(text_displayed)


driver.find_element(By.XPATH,"//button[text()='Try it']").click()
alert=driver.switch_to.alert
alert.send_keys("harry")
alert.accept()

text_displayed=driver.find_element(By.ID,"demo").text
print(text_displayed)

driver.quit()
```
