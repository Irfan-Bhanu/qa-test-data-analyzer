```python
#POST - to create data, to send data to server to create new resource and success code is 201

import requests
url = "https://jsonplaceholder.typicode.com/posts"

payload = {

    'user': 'Harry',
    'title': 'Test title',
    'body':'This is for test',
    'userid': '1'
}


```

```python
response = requests. post(url,json=payload)

print(response.status_code)
print(response.json())

data=response.json()
print('created id',data['id'])

if response.status_code==201:
    print("POST success")
else:
    print('post failed')



if data['title'] =='Test title':
    print('Title validation passed')
else:
    print('Title validation failed')
```

```python
#Put- to update existing resource and #delete- remove resources

import requests

url = "https://jsonplaceholder.typicode.com/posts/1"

payload = {
    "id": 101,
    'user': 'Harry',
    'title': 'Test title',
    'body':'updated body test',
    'userid': '2'
}

response = requests.put(url, json=payload)

print(response.status_code)
print(response.json())

data = response.json()

if response.status_code == 200:
    print("PUT Success")

if data["body"] == "updated body test":
    print("Update Verified")
```

```python
#delete is to delete the resource - DELETE request removes a specific resource identified by the URL. After deletion, accessing the same resource should return a 404 response

import requests

url = "https://jsonplaceholder.typicode.com/posts/1"

response = requests.delete(url)

print(response.status_code)

if response.status_code in [200, 204]:
    print("Delete Successful")

response = requests.get('https://jsonplaceholder.typicode.com/posts/1')
print(response.status_code)
```

```python
#API+data validation

import requests

response =requests.get("https://jsonplaceholder.typicode.com/posts")

print(response.status_code)
data = response.json()

print(len(data))

#check missing values

for i in data:
    if i['title'] == ' ' and  i['body'] == ' ':
        print("missing values at: ", i['id'])
    else:
        print("No missing values")
```

```python
#check data types: we can use isinstance or type

for i in data:
    if not isinstance(i['id'],int):
        print('invalid data type')
    else:
        print(f" id:{i['id']} --> type: {type(i['id'])}")
```

```python
#check duplicate ids

ids=[]

for i in data:
    if i['id'] in ids:
        print("duplicate found:" ,i['id'])
    else:
        ids.append(i['id'])
```
