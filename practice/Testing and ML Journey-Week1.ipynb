```python
#1. Store a test results and count no of pass and fail count
test_results = ["pass", "fail", "pass", "fail","pass"]

print("-------------Test Summary Report--------------")

total =len(test_results)
print(f"The total count is {total}")

pass_count=0
fail_count=0

for i in test_results:
    if i=='pass': 
        pass_count += 1
    else:
        fail_count+=1
print(f"Pass count is {pass_count}")
print(f"fail count is {fail_count}")
pass_per = (pass_count/total)*100
fail_per=(fail_count/total)*100
print(f"total pass percentage is {pass_per:.2f}")
print(f"total fail percentage is {fail_per:.2f}")

if fail_per > 50:
    print("High Risk")
elif fail_per > 20:
    print("Moderate Risk")
else:
    print("Low Risk")
```

```python
#write test function
def test_summary(test_results):

    
    total = len(test_results)
    
    failed_module =[]
    pass_count=0
    fail_count=0
    for key,value in test_results.items():
        if value == "fail":
            #print(f"failed module is {key}")
            fail_count+=1
            failed_module.append(key)
            
        else:
            pass_count+=1
            
            
    #print(failed_module)
    #print(f"failed count is {fail_count}")
    #print(f"pass count is {pass_count}")
    
    pass_per = (pass_count/total)*100
    fail_per=(fail_count/total)*100
    #print(f"total pass percentage is {pass_per:.2f}")
    #print(f"total fail percentage is {fail_per:.2f}")
    
    if fail_per > 50:
        print("Status : Critical")
    elif 30 <= fail_per <=50:
        print("Status : Warning")
    else:
        print("Status : Stable")

    #return pass_count,fail_count,fail_per,pass_per,failed_module
    return {
    "pass": pass_count,
    "fail": fail_count,
    "fail_per": fail_per,
    "failed_modules": failed_module
        }


```

```python
#use strip and lower to clear messy data and #Read values from a excel file and pass it into the function without pandas.convert excel to csv
import pandas as pd
import csv

# Step 1: Read Excel file
excel_file = r"C:\Users\bhanu\Downloads\messy_test_results.xlsx"
data = pd.read_excel(excel_file)

# Step 2: Convert to CSV
csv_file = r"C:\Users\bhanu\Downloads\test_results.csv"
data.to_csv(csv_file, index=False)

print("Excel converted to CSV successfully!")

# Step 3: Read CSV using csv module
data_dict = {}

with open(csv_file, mode='r') as file:
    reader = csv.DictReader(file)

    for row in reader:
        # Step 4: Clean data
        module = row["module"].strip()
        result = row["result"].strip().lower()

        # Filter valid values
        if result in ('pass', 'fail'):
            data_dict[module] = result

# Step 5: Pass into your function
result = test_summary(data_dict)

print("------ Final Output ------")
print(result)
```
